````python
%%writefile my_program.cpp

//START YOUR CODE HERE
///////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////////
#include<iostream>
using namespace std;

int main(){
    int numOfSubjects=3;
    float grade[3];

    //input section
    cout<<"請輸入第一科成績: ";
    cin>>grade[0];
    cout<<"請輸入第二科成績: ";
    cin>>grade[1];
    cout<<"請輸入第三科成績: ";
    cin>>grade[2];

    //find average
    float average=0;
    for(int i=0;i<numOfSubjects;i++){
        average+=grade[i];
    }
    average/=numOfSubjects;

    //find evaluation result
    string evaluation;
    if (average>=90){
        evaluation="優秀";
    }
    else if (average>=70){
        evaluation="良好";
    }
    else if (average>=60){
        evaluation="及格";

    }
    else{
        evaluation="不及格";
    }


    cout<<"平均分數是 "<<average<<"，評估結果:"<<evaluation;
    return 0;
}

````

````python
#include <bits/stdc++.h>


using namespace std;

int mov_r, mov_d, N, M;
/*
the count of move to right when moving to the normal pos;
the count of move to down when moving to the normal pos;
the screen has N rows;
the screen has M columns;
this is set as a global variable because it will be used several times in the functions
*/

void right(vector<vector<char> >& arr) {
    for (int i = 0; i < ::N; i++) {
    	int first_vl=arr[i][0]; //create a temp for the fist value since it will be wiped out in the following program
        for(int j=0; j< ::M; j++){
        	if(M==1||M==0||j==0){	} //do nothing when there is only one column, zero column, or if we are searching for the first value(we will use first_value to assign a value to it instred of doing it here

			else{
				arr[i][j]=arr[i][j-1];
			}

		}
		arr[i][0]=first_vl;
    }
}

void down(vector<vector<char> >& arr) {
    for (int i = 0; i < ::N; i++) {
    	int first_vl=arr[0][i]; //create a temp for the fist value since it will be wiped out in the following program
        for(int j=0; j< ::M; j++){
        	if(N==1||N==0||i==0){	} //do nothing when there is only one row, zero row, or if we are searching for the first value(we will use first_value to assign a value to it instred of doing it here

			else{
				arr[j][i]=arr[j-1][i];
			}

		}
		arr[0][i]=first_vl;
    }
}

bool match_ch_row(vector<vector<char> >&arr_ab, vector<vector<char> >& arr_nor){
	//add the acii codes to check
	for(int i=0; i< ::M; i++){
		int nor[ ::M]={0};
		int ab[ ::M]={0}; //these will record the sum of each row
		for(int j=0; i< ::N; j++){
			nor[i]+=(int)arr_nor[i][j];
		}
		for(int j=0; i< ::N; j++){
			ab[i]+=(int)arr_ab[i][j];
		}
		if(nor[i]!=ab[i]){
			return false;
		}
	}
	return true;

}

bool match_ch_col(vector<vector<char> >&arr_ab, vector<vector<char> >& arr_nor){
	//add the acii codes to check
	for(int i=0; i< ::M; i++){
		int nor[ ::M]={0};
		int ab[ ::M]={0}; //these will record the sum of each row
		for(int j=0; i< ::N; j++){
			nor[i]+=(int)arr_nor[j][i];
		}
		for(int j=0; i< ::N; j++){
			ab[i]+=(int)arr_ab[j][i];
		}
		if(nor[i]!=ab[i]){
			return false;
		}
	}
	return true;


}


int main(){
	//input

	cin>>N>>M;

	//input normal screen pos : a
	vector<vector<char> > a(N, vector<char>(M));
	for(int i=0; i<N; i++){
		for(int j=0; j<M; j++){
			cin>>a[i][j];
		}
	}
	//input error screen pos : b
	vector<vector<char> > b(N, vector<char>(M));

	for(int i=0; i<N; i++){
		for(int j=0; j<M; j++){
			cin>>b[i][j];
		}
	}

	//processing
	/*思路:
	THERE ARE ONLY TWO WAYS TO MOVE THE BLOCKS, AND THERE ARE THEREFORE ONLY TWO SOLUSIONS FOR EVERY ARRAY
	1. 原b陣列先往右對應好row，數次數x_1 ，成 b'
	2. 新 b'往下對應好column，數次數y_1 ，成 b''
	3.sol_1=x+y

	4. 原b陣列先往下對應好column，數次數x_2 ，成 b'
	5. 新 b'往右對應好row，數次數y_2 ，成 b''
	6.sol_2=k+p*/

	int sol_1=0, x_1=0, y_1=0;
	int sol_2=0, x_2=0, y_2=0;

	//vector<vector<char> > temp(N, vector<char>(M)) = b;
	vector<vector<char> > temp = b;

	for(int i=0; i<M; i++){
		right(temp);
		x_1++;
		if(temp==a){
			break;
		}
	}
	for(int i=0; i<N; i++){
		down(b);
		y_1++;
		if(b==a){
			break;
		}
	}
	sol_1=x_1+y_1;

	//output


	for(int i=0; i<M; i++){
		down(b);
		x_1++;
		if(match_ch_row(b,a)){
			break;
		}
	}
	for(int i=0; i<N; i++){
		right(b);
		y_1++;
		if(match_ch_col(b, a)){
			break;
		}
	}
	sol_2=x_2+y_2;

	if(sol_1<sol_2){
		cout<<x_1<<x_2;
	}
	else{
		cout<<y_1<<y_2;
	}

}
````

````python
%%writefile my_program.cpp

#include <iostream>
#include <vector>
#include <algorithm>
#include <numeric>

using namespace std;

int main() {
    // 1. I/O 加速
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n;
    if (!(cin >> n)) return 0;

    // 2. 一次性配置好所有記憶體，避免動態擴充的開銷
    vector<vector<int>> scores(n);
    for (int i = 0; i < n; ++i) {
        int k;
        cin >> k;
        scores[i].resize(k); // 直接開好空間
        for (int j = 0; j < k; ++j) {
            cin >> scores[i][j];
        }
    }

    int x, y;
    cin >> x >> y;

    int max_score = -1;
    int best_id = 0;

    // 3. 原地處理：排序、加總、找最大值一次完成
    for (int i = 0; i < n; ++i) {
        // 原地排序該學生的成績
        sort(scores[i].begin(), scores[i].end());

        int count = scores[i].size() - x - y;
        if (count <= 0) continue; // 避免無效資料導致除以零

        // 利用 std::accumulate 快速加總指定範圍，避開迴圈手寫
        /*
        accumulate 語法介紹:
        #include <numeric> // Required for std::accumulate
        std::accumulate(first, last, init, op);

        first: Iterator to the first element of the range.
        last: Iterator to the element one after the last element of the range.
        init: Initial value to start the accumulation with.
        op (optional): Binary operation function or functor. Defaults to addition (+)(可以寫multiply...)
        */
        // 起點：begin() + y (跳過最低的 y 個)
        // 終點：end() - x (捨去最高的 x 個)
        int sum = accumulate(scores[i].begin() + y, scores[i].end() - x, 0);
        int avg = sum / count;

        // 邊計算邊更新最大值，省去額外存 avg 陣列的空間與時間
        if (avg > max_score) {
            max_score = avg;
            best_id = i + 1; // 假設學生編號從 1 開始
        }
    }

    cout << best_id << " " << max_score << "\n";

    return 0;
}
````

### 第二題：pB — 花圃選區

**題意說明**
有一個 $N \times M$ 的花圃，左上角座標為 $(0, 0)$。每個格子的值 $A_{i,j}$ 代表：
* $-1$：這個位置不能種花（障礙物）。
* 正整數：這個位置種了第 $A_{i,j}$ 種花。

校長想挑選一個 $R \times C$ 大小的矩形區域，必須同時滿足以下三個條件（依優先順序）：
1. 區域內每個位置都不能是 $-1$（全部都是能種花的格子）。
2. 在這個前提下，區域內的花種類數要越多越好（不同正整數的個數）。
3. 如果仍有多個區域花種類數相同，則選左上角座標最靠近 $(0, 0)$ 的那個（以曼哈頓距離衡量：$i + j$）。若曼哈頓距離也相同，則選行座標 $(i)$ 較小的；若 $i$ 也相同，則選列座標 $(j)$ 較小的。

最後輸出：最佳區域的左上角座標，以及該區域內的花種類數。

**輸入格式**

N  M  R  C

A_{0,0} ... A_{0,M-1}

...

A_{N-1,0} ... A_{N-1,M-1}

````python
#include<iostream>
#include<vector>
#include <algorithm>
using namespace std;

int findScore(vector<vector<int> >toFind, int row, int col){
	//criteria: if there is any -1 in the block: return 0; if there is no -1, find the score by adding how many sorts of flower an area has
	vector<int>  numOfFl;
	for(int i=0; i<row; i++){
		toFind[i].push_back(0);
		//push a 0 to every end of the row so that if a -1 is found in the end of the line, we can know it
		for(int j=0; j<col; j++){
			int temp=toFind[i][j];
			if (temp==-1) return -1;
			numOfFl[toFind[i][j]]++;
		}
	}

	//find the amount of index in which their value isn't 0
	int max_search=numOfFl.size();
	int retN=0;
	for(int i=0; i<max_search; i++){
		if(numOfFl[i]>0) retN++;
	}

	return retN;
}


int main(){
	ios_base::sync_with_stdio(false);
	cin.tie(NULL);
	//define and input
	int  N,M,R,C=0;
	cin>>N>>M>>R>>C;

	vector<vector<int> > flmap;
	//input this map
	for(int i=0; i<N; i++){
		for(int j=0; j<M; j++){
			cin>>flmap[i][j];
		}
	}



	vector<vector<int> > scmap;//a map of the score in flmap in its range of RxC
	//build  scmap
	for(int i=0; i<N-R; i++){
		for(int j=0; j<M-C; j++){
        	vector<vector<int> > temp;

        	//build temp
        	for(int n=0; n<R;n++){
        		for(int m=0; m<C; m++){
        			temp[n][m]=flmap[i+n][j+m];
				}
			}

			scmap[i][j]=findScore(temp,R,C);
		}
	}

	//select the best pos in the scmap

	int best_i = -1;
    int best_j = -1;
    int max_kinds = -1;


 for (int i = 0; i <= N - R; i++) {
        for (int j = 0; j <= M - C; j++) {
            // 如果這個區域內有 -1，scmap[i][j] 會是 -1，直接跳過
            if (scmap[i][j] == -1) continue;

            // 如果是第一個找到的合法區域，直接當作目前的最佳解
            if (best_i == -1) {
                best_i = i;
                best_j = j;
                max_kinds = scmap[i][j];
                continue;
            }

            // 比較優先順序
            bool should_update = false;

            // 優先順序 1：種類數越多越好
            if (scmap[i][j] > max_kinds) {
                should_update = true;
            }
            // 種類數相同時的 Tie-breakers
            else if (scmap[i][j] == max_kinds) {
                int current_dist = i + j;
                int best_dist = best_i + best_j;

                // 優先順序 2：曼哈頓距離越小越好
                if (current_dist < best_dist) {
                    should_update = true;
                } else if (current_dist == best_dist) {
                    // 優先順序 3：列座標 (i) 越小越好
                    if (i < best_i) {
                        should_update = true;
                    } else if (i == best_i) {
                        // 優先順序 4：行座標 (j) 越小越好
                        if (j < best_j) {
                            should_update = true;
                        }
                    }
                }
            }

            if (should_update) {
                best_i = i;
                best_j = j;
                max_kinds = scmap[i][j];
            }
        }
    }

    // 輸出結果：最佳區域的左上角座標，以及該區域內的花種類數
    cout << best_i << " " << best_j << " " << max_kinds << "\n";

    return 0;	}

````

````python
%%writefile my_program.cpp
//ai生成的ac版本

#include <iostream>
#include <string>
#include <vector>

using namespace std;

int N;

// 負責計算最終產生的算式字串是否等於 0
bool evaluate(string expr) {
    long long total = 0;
    long long current_num = 0;
    char last_op = '+'; // 預設第一個數字前面是加號

    for (int i = 0; i < expr.length(); i++) {
        char c = expr[i];
        if (c >= '0' && c <= '9') {
            current_num = current_num * 10 + (c - '0');
        } else if (c == '+' || c == '-') {
            // 遇到新的加減號，先把上一個數字結算
            if (last_op == '+') total += current_num;
            else total -= current_num;

            current_num = 0;
            last_op = c;
        }
        // 如果是空格 ' '，就直接跳過，因為下一輪迴圈會自動把數字接在 current_num 後面
    }
    // 結算最後一個數字
    if (last_op == '+') total += current_num;
    else total -= current_num;

    return total == 0;
}

// idx: 當前要放入的數字
// expr: 目前累積的算式字串
void dfs(int idx, string expr) {
    if (idx > N) {
        if (evaluate(expr)) {
            cout << expr << "\n";
        }
        return;
    }

    // 嚴格按照 ASCII 順序：' ', '+', '-'
    dfs(idx + 1, expr + " " + to_string(idx));
    dfs(idx + 1, expr + "+" + to_string(idx));
    dfs(idx + 1, expr + "-" + to_string(idx));
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    while (cin >> N) {
        // 從數字 2 開始，初始算式字串為 "1"
        dfs(2, "1");
        cout << "\n";
    }
    return 0;
}
````

````python
%%writefile my_program.cpp

#include <iostream>
#include <queue> // 引入佇列，用來做 BFS 排隊

using namespace std;

int main() {
    int R, C;

    // 1. 讀取迷宮的大小
    // cin 是 C++ 讀取輸入的指令
    if (!(cin >> R >> C)) return 0;

    int grid[25][25]; // 用來存地圖，題目說最大 20x20，我們開 25 比較安全
    int steps[25][25]; // 用來記錄走到每一格總共經過了幾個黑格

    // 2. 讀取迷宮資料，並初始化步數陣列
    for (int i = 0; i < R; i++) {
        for (int j = 0; j < C; j++) {
            cin >> grid[i][j];
            steps[i][j] = -1; // -1 代表這一格還沒走過
        }
    }

    // 3. 準備 BFS (水波蔓延)
    // queue 裡面存的是 pair<int, int>，也就是 (x 座標, y 座標)
    queue<pair<int, int>> q;

    // 把起點放進排隊隊伍中
    q.push({0, 0});
    steps[0][0] = 1; // 起點本身就算 1 個黑格

    // 定義四個方向：上、下、左、右
    // x 改變，y 不變：dx 陣列
    // x 不變，y 改變：dy 陣列
    int dx[] = {-1, 1, 0, 0};
    int dy[] = {0, 0, -1, 1};

    // 4. 開始探索，直到排隊隊伍空了為止
    while (!q.empty()) {
        // 取出隊伍最前面的人 (目前的格子)
        int current_x = q.front().first;
        int current_y = q.front().second;
        q.pop(); // 拿出來之後，就讓他離開隊伍

        // 檢查這個格子的上下左右 4 個方向
        for (int i = 0; i < 4; i++) {
            int next_x = current_x + dx[i];
            int next_y = current_y + dy[i];

            // 檢查下一步是否合法：
            // (1) next_x 不能超出邊界 0 ~ R-1
            // (2) next_y 不能超出邊界 0 ~ C-1
            // (3) grid[next_x][next_y] 必須是 1 (黑格)
            // (4) steps[next_x][next_y] 必須是 -1 (還沒走過)
            if (next_x >= 0 && next_x < R &&
                next_y >= 0 && next_y < C &&
                grid[next_x][next_y] == 1 &&
                steps[next_x][next_y] == -1) {

                // 如果可以走，步數就是「現在這格的步數 + 1」
                steps[next_x][next_y] = steps[current_x][current_y] + 1;

                // 把新的格子加入排隊隊伍，等待之後探索它的鄰居
                q.push({next_x, next_y});
            }
        }
    }

    // 5. 輸出終點的步數
    cout << steps[R - 1][C - 1] << "\n";

    return 0;
}
````

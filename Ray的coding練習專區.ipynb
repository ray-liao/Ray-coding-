### 第一題：pA — 學生成績計算

**題意說明**
有 $N$ 位學生，每位學生參加了若干次練習賽。
* 第 $i$ 位學生參加了 $K_i$ 次練習賽。
* 第 $i$ 位學生第 $j$ 次的成績記為 $s_{i,j}$ ($0 \le s_{i,j} \le 100$)。

**總成績的算法**
把該學生所有成績去掉最高的 $x$ 次和最低的 $y$ 次後，剩下的成績取平均（無條件捨去，即整數除法）。

**題目要求**
找出總成績最高的那位學生，輸出他的學生編號以及總成績。

**輸入格式**

N

K_1  s_1,1   s_1,2  ...  s_1

K_2  s_2,1   s_2,2  ...  s_2

...

K_N  s_N,1   s_N,2  ...  s_N

x    y

````python
%%測資

3
5 10 20 30 40 50
6 80 85 90 95 100 100
4 100 0 100 0
1 1

return:
2 92
````

````python
%%writefile my_program.cpp

#include<iostream>
#include <vector>
#include <array>
#include <algorithm>

int sortAD(std::vector<int> TBSorted,int noMa, int nomi){
  std::sort(TBSorted.begin(), TBSorted.end());//sort

  //find avg excluding unwanted min and max
  int sum=0;
  for(int i=nomi; i<(TBSorted.size())-noMa; i++){
    sum+=TBSorted[i];
  }
  return (sum/((TBSorted.size())-noMa-nomi));
}

int main(){
  //define
  std::ios_base::sync_with_stdio(false);
  std::vector<std::vector<int>> numbers;
  int numStudent=0;
  int miMax, miMin;

  //input
  std::cin>>numStudent;
  numbers.resize(numStudent);
  for(int i=0; i<numStudent; i++){
    int numSc=0;
    for(int j=0; i<numSc; j++){
      std::cin >> numbers[i][j];
    }
  }
  std::cin>>miMax>>miMin;

  //delete max and min and find average
  int avg[numStudent]={0};
  for(int i=0; i<numStudent; i++){
    std::vector<int> TBSorted;
    for(int j=1; j<numbers[i][0];j++){
      TBSorted[j-1]=numbers[i][j];//assign to a new vector
    }
    avg[i]=sortAD(TBSorted,miMax,miMin);
  }

  //find max and its index num
  int maxScore=0;
  int index=0;
  for(; index<numStudent; ++index){
    maxScore=(avg[index]>maxScore? avg[index]:maxScore);
  }
  std::cout<<"Student with max score: "<<index<<"\n With the score of: "<<maxScore;

}
````

````python
%%writefile my_program.cpp
#include<iostream>
#include <vector>
#include <algorithm>

int sortAD(std::vector<int> TBSorted,int noMa, int nomi){
  std::sort(TBSorted.begin(), TBSorted.end());//sort

  //find avg excluding unwanted min and max
  int sum=0;
  for(int i=nomi; i<(TBSorted.size())-noMa; i++){
    sum+=TBSorted[i];
  }
 // 計算有效成績的數量，並回傳平均 (避免除以 0 的保護機制)
    int validCount = TBSorted.size() - noMa - nomi;
    if (validCount <= 0) return 0;

    return sum / validCount;}

int main(){
	// 優化輸入輸出速度
	std::ios_base::sync_with_stdio(false);
	std::cin.tie(NULL);
	//define
	int numStudent=0;
	int miMax, miMin;

  //input
	std::cin>>numStudent;
	std::vector<std::vector<int> > numbers(numStudent);
  for(int i=0; i<numStudent; i++){
    int numSc=0;
    std::cin >> numSc; // 讀取該名學生有幾個成績 (K_i)
    for(int j=0; j<numSc; j++){
    	int score;
    	std::cin >> score;
    	numbers[i].push_back(score); // 使用 push_back 動態新增成績
    	//***重要語法**//
    }
  }
  std::cin>>miMax>>miMin;

  //delete max and min and find average
  int avg[numStudent]={0};
  for (int i = 0; i < numStudent; i++) {
        // 直接將 numbers[i] 傳入 sortAD
        // 因為 sortAD 的參數是傳值 (pass by value)，所以會自動產生一份拷貝來排序
        avg[i] = sortAD(numbers[i], miMax, miMin);//numbers[i]可以直接傳一行
    }
  //find max and its index num
    int maxIndex = 0;
    int maxScore = -1;
    for (int i = 0; i < numStudent; i++) {
        if (avg[i] > maxScore) {
            maxScore = avg[i];
            maxIndex = i; // 紀錄最高分是哪位學生
        }
    }
  std::cout<<"Student with max score: "<<maxIndex+1<<"\nWith the score of: "<<maxScore;

}
````

````python
%%writefile my_program.cpp

#include <iostream>
#include <vector>
#include <algorithm>
#include <numeric>

using namespace std;

int main() {
    // 1. I/O 加速
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    int n;
    if (!(cin >> n)) return 0;

    // 2. 一次性配置好所有記憶體，避免動態擴充的開銷
    vector<vector<int>> scores(n);
    for (int i = 0; i < n; ++i) {
        int k;
        cin >> k;
        scores[i].resize(k); // 直接開好空間
        for (int j = 0; j < k; ++j) {
            cin >> scores[i][j];
        }
    }

    int x, y;
    cin >> x >> y;

    int max_score = -1;
    int best_id = 0;

    // 3. 原地處理：排序、加總、找最大值一次完成
    for (int i = 0; i < n; ++i) {
        // 原地排序該學生的成績
        sort(scores[i].begin(), scores[i].end());

        int count = scores[i].size() - x - y;
        if (count <= 0) continue; // 避免無效資料導致除以零

        // 利用 std::accumulate 快速加總指定範圍，避開迴圈手寫
        /*
        accumulate 語法介紹:
        #include <numeric> // Required for std::accumulate
        std::accumulate(first, last, init, op);

        first: Iterator to the first element of the range.
        last: Iterator to the element one after the last element of the range.
        init: Initial value to start the accumulation with.
        op (optional): Binary operation function or functor. Defaults to addition (+)(可以寫multiply...)
        */
        // 起點：begin() + y (跳過最低的 y 個)
        // 終點：end() - x (捨去最高的 x 個)
        int sum = accumulate(scores[i].begin() + y, scores[i].end() - x, 0);
        int avg = sum / count;

        // 邊計算邊更新最大值，省去額外存 avg 陣列的空間與時間
        if (avg > max_score) {
            max_score = avg;
            best_id = i + 1; // 假設學生編號從 1 開始
        }
    }

    cout << best_id << " " << max_score << "\n";

    return 0;
}
````

### 第二題：pB — 花圃選區

**題意說明**
有一個 $N \times M$ 的花圃，左上角座標為 $(0, 0)$。每個格子的值 $A_{i,j}$ 代表：
* $-1$：這個位置不能種花（障礙物）。
* 正整數：這個位置種了第 $A_{i,j}$ 種花。

校長想挑選一個 $R \times C$ 大小的矩形區域，必須同時滿足以下三個條件（依優先順序）：
1. 區域內每個位置都不能是 $-1$（全部都是能種花的格子）。
2. 在這個前提下，區域內的花種類數要越多越好（不同正整數的個數）。
3. 如果仍有多個區域花種類數相同，則選左上角座標最靠近 $(0, 0)$ 的那個（以曼哈頓距離衡量：$i + j$）。若曼哈頓距離也相同，則選行座標 $(i)$ 較小的；若 $i$ 也相同，則選列座標 $(j)$ 較小的。

最後輸出：最佳區域的左上角座標，以及該區域內的花種類數。

**輸入格式**

N  M  R  C

A_{0,0} ... A_{0,M-1}

...

A_{N-1,0} ... A_{N-1,M-1}

````python
#include<iostream>
#include<vector>
#include <algorithm>
using namespace std;

int findScore(vector<vector<int> >toFind, int row, int col){
	//criteria: if there is any -1 in the block: return 0; if there is no -1, find the score by adding how many sorts of flower an area has
	vector<int>  numOfFl;
	for(int i=0; i<row; i++){
		toFind[i].push_back(0);
		//push a 0 to every end of the row so that if a -1 is found in the end of the line, we can know it
		for(int j=0; j<col; j++){
			int temp=toFind[i][j];
			if (temp==-1) return -1;
			numOfFl[toFind[i][j]]++;
		}
	}

	//find the amount of index in which their value isn't 0
	int max_search=numOfFl.size();
	int retN=0;
	for(int i=0; i<max_search; i++){
		if(numOfFl[i]>0) retN++;
	}

	return retN;
}


int main(){
	ios_base::sync_with_stdio(false);
	cin.tie(NULL);
	//define and input
	int  N,M,R,C=0;
	cin>>N>>M>>R>>C;

	vector<vector<int> > flmap;
	//input this map
	for(int i=0; i<N; i++){
		for(int j=0; j<M; j++){
			cin>>flmap[i][j];
		}
	}



	vector<vector<int> > scmap;//a map of the score in flmap in its range of RxC
	//build  scmap
	for(int i=0; i<N-R; i++){
		for(int j=0; j<M-C; j++){
        	vector<vector<int> > temp;

        	//build temp
        	for(int n=0; n<R;n++){
        		for(int m=0; m<C; m++){
        			temp[n][m]=flmap[i+n][j+m];
				}
			}

			scmap[i][j]=findScore(temp,R,C);
		}
	}

	//select the best pos in the scmap

	int best_i = -1;
    int best_j = -1;
    int max_kinds = -1;


 for (int i = 0; i <= N - R; i++) {
        for (int j = 0; j <= M - C; j++) {
            // 如果這個區域內有 -1，scmap[i][j] 會是 -1，直接跳過
            if (scmap[i][j] == -1) continue;

            // 如果是第一個找到的合法區域，直接當作目前的最佳解
            if (best_i == -1) {
                best_i = i;
                best_j = j;
                max_kinds = scmap[i][j];
                continue;
            }

            // 比較優先順序
            bool should_update = false;

            // 優先順序 1：種類數越多越好
            if (scmap[i][j] > max_kinds) {
                should_update = true;
            }
            // 種類數相同時的 Tie-breakers
            else if (scmap[i][j] == max_kinds) {
                int current_dist = i + j;
                int best_dist = best_i + best_j;

                // 優先順序 2：曼哈頓距離越小越好
                if (current_dist < best_dist) {
                    should_update = true;
                } else if (current_dist == best_dist) {
                    // 優先順序 3：列座標 (i) 越小越好
                    if (i < best_i) {
                        should_update = true;
                    } else if (i == best_i) {
                        // 優先順序 4：行座標 (j) 越小越好
                        if (j < best_j) {
                            should_update = true;
                        }
                    }
                }
            }

            if (should_update) {
                best_i = i;
                best_j = j;
                max_kinds = scmap[i][j];
            }
        }
    }

    // 輸出結果：最佳區域的左上角座標，以及該區域內的花種類數
    cout << best_i << " " << best_j << " " << max_kinds << "\n";

    return 0;	}

````

````python
%%writefile my_program.cpp
//ai生成的ac版本

#include <iostream>
#include <string>
#include <vector>

using namespace std;

int N;

// 負責計算最終產生的算式字串是否等於 0
bool evaluate(string expr) {
    long long total = 0;
    long long current_num = 0;
    char last_op = '+'; // 預設第一個數字前面是加號

    for (int i = 0; i < expr.length(); i++) {
        char c = expr[i];
        if (c >= '0' && c <= '9') {
            current_num = current_num * 10 + (c - '0');
        } else if (c == '+' || c == '-') {
            // 遇到新的加減號，先把上一個數字結算
            if (last_op == '+') total += current_num;
            else total -= current_num;

            current_num = 0;
            last_op = c;
        }
        // 如果是空格 ' '，就直接跳過，因為下一輪迴圈會自動把數字接在 current_num 後面
    }
    // 結算最後一個數字
    if (last_op == '+') total += current_num;
    else total -= current_num;

    return total == 0;
}

// idx: 當前要放入的數字
// expr: 目前累積的算式字串
void dfs(int idx, string expr) {
    if (idx > N) {
        if (evaluate(expr)) {
            cout << expr << "\n";
        }
        return;
    }

    // 嚴格按照 ASCII 順序：' ', '+', '-'
    dfs(idx + 1, expr + " " + to_string(idx));
    dfs(idx + 1, expr + "+" + to_string(idx));
    dfs(idx + 1, expr + "-" + to_string(idx));
}

int main() {
    ios_base::sync_with_stdio(false);
    cin.tie(NULL);

    while (cin >> N) {
        // 從數字 2 開始，初始算式字串為 "1"
        dfs(2, "1");
        cout << "\n";
    }
    return 0;
}
````

````python
%%writefile my_program.cpp

#include <iostream>
#include <queue> // 引入佇列，用來做 BFS 排隊

using namespace std;

int main() {
    int R, C;

    // 1. 讀取迷宮的大小
    // cin 是 C++ 讀取輸入的指令
    if (!(cin >> R >> C)) return 0;

    int grid[25][25]; // 用來存地圖，題目說最大 20x20，我們開 25 比較安全
    int steps[25][25]; // 用來記錄走到每一格總共經過了幾個黑格

    // 2. 讀取迷宮資料，並初始化步數陣列
    for (int i = 0; i < R; i++) {
        for (int j = 0; j < C; j++) {
            cin >> grid[i][j];
            steps[i][j] = -1; // -1 代表這一格還沒走過
        }
    }

    // 3. 準備 BFS (水波蔓延)
    // queue 裡面存的是 pair<int, int>，也就是 (x 座標, y 座標)
    queue<pair<int, int>> q;

    // 把起點放進排隊隊伍中
    q.push({0, 0});
    steps[0][0] = 1; // 起點本身就算 1 個黑格

    // 定義四個方向：上、下、左、右
    // x 改變，y 不變：dx 陣列
    // x 不變，y 改變：dy 陣列
    int dx[] = {-1, 1, 0, 0};
    int dy[] = {0, 0, -1, 1};

    // 4. 開始探索，直到排隊隊伍空了為止
    while (!q.empty()) {
        // 取出隊伍最前面的人 (目前的格子)
        int current_x = q.front().first;
        int current_y = q.front().second;
        q.pop(); // 拿出來之後，就讓他離開隊伍

        // 檢查這個格子的上下左右 4 個方向
        for (int i = 0; i < 4; i++) {
            int next_x = current_x + dx[i];
            int next_y = current_y + dy[i];

            // 檢查下一步是否合法：
            // (1) next_x 不能超出邊界 0 ~ R-1
            // (2) next_y 不能超出邊界 0 ~ C-1
            // (3) grid[next_x][next_y] 必須是 1 (黑格)
            // (4) steps[next_x][next_y] 必須是 -1 (還沒走過)
            if (next_x >= 0 && next_x < R &&
                next_y >= 0 && next_y < C &&
                grid[next_x][next_y] == 1 &&
                steps[next_x][next_y] == -1) {

                // 如果可以走，步數就是「現在這格的步數 + 1」
                steps[next_x][next_y] = steps[current_x][current_y] + 1;

                // 把新的格子加入排隊隊伍，等待之後探索它的鄰居
                q.push({next_x, next_y});
            }
        }
    }

    // 5. 輸出終點的步數
    cout << steps[R - 1][C - 1] << "\n";

    return 0;
}
````
